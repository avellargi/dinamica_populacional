## 3. Análise e Discussão de Diferentes Modelos Populacionais

<a id="malthus"></a>
### 3.1 Modelo de Malthus

$ \ \ \ \ \ $ Uma das primeiras tentativas de modelagem do crescimento populacional, por meio da
matemática, foi feita pelo economista inglês Thomas Malthus em 1798. O crescimento populacional exponencial descreve uma situação ideal em que uma população aumenta continuamente a uma taxa proporcional ao seu tamanho. Nesse modelo, a taxa de crescimento por indivíduo (taxa per capita) permanece constante ao longo do tempo, independentemente do tamanho populacional. [[7]](#1) <br>
Isso implica que, quanto maior a população, maior será o número absoluto de novos indivíduos gerados, levando a um crescimento cada vez mais acelerado. Matematicamente, esse comportamento é representado pela equação diferencial:

$$
\frac{dN}{dt} = rN
$$
onde:<br>
$ N $: representa o tamanho da população <br>
$ r $: é a taxa de crescimento intrínseca da população <br>

A solução dessa equação, sujeita a condição de população inicial $ N(0) = N_0 $, é dada por:

$$
N(t) = N_0 e^{rt}
$$

$ \ \ \ \ \ $ Vale ressaltar que esse modelo pressupõe condições ideais, como disponibilidade ilimitada de recursos e ausência de competição, sendo considerado uma aproximação válida apenas em estágios iniciais de crescimento populacional, como na colonização de um novo ambiente ou em culturas de microrganismos em laboratório.

```python
def crescimento_exponencial(r, N0, tf=100, pontos=500):
    """
    Modelo de crescimento exponencial de uma população (Malthus).

    Argumentos:
    r : float - Taxa de crescimento da população.
    N0 : float - População inicial no tempo t = 0.
    tf : float (opcional) - Tempo final da simulação (padrão = 100).
    pontos : int (opcional) - Número de pontos no intervalo de tempo (padrão = 500).

    Retorno: uma tupla contendo t, N
    t : array (numpy) - Vetor de tempo gerado para a simulação.
    N : array (numpy) - Tamanho da pop. ao longo do tempo.
    
    """
    t = np.linspace(0, tf, pontos)
    
    N = N0 * np.exp(r * t)
    
    resultado_exp = (t, N)
    
    return resultado_exp
```

$ \ \ \ \ \ $ A partir da simulação, foi gerado o gráfico da população ao longo do tempo, permitindo a análise do modelo exponencial para diferentes valores dos parâmetros. Por meio da interface interativa, é possível variar a taxa de crescimento $ r $ e a população inicial $ N_0 $, observando como essas alterações influenciam diretamente o comportamento da curva. Em particular, a variação de $ N_0 $ modifica o ponto inicial do gráfico, enquanto mudanças em $ r $ alteram a inclinação da curva, evidenciando seu impacto na velocidade de crescimento. 

```python
def plot_cresc_exp(r, N0):
    resultado_exp = crescimento_exponencial(r, N0)

    plt.plot(*resultado_exp, label="Crescimento Exponencial")
    plt.title("Modelo de Malthus")
    plt.xlabel("Tempo")
    plt.ylabel("População")
    plt.ylim(0, 350)
    plt.legend()
    plt.grid()
    plt.show()

interact(
    plot_cresc_exp,
    r = FloatSlider(value=0.05, min=0.001, max=1, step=0.001, description='Taxa de crescimento', style={'description_width':'200px'}, layout=Layout(width='600px')),
                    
    N0 = FloatSlider(value=10, min=0, max=200, step=0.001, description='População Inicial', style={'description_width': '200px'}, layout=Layout(width='600px'))
);
```

$ \ \ \ \ \ $ Observa‑se que o comportamento da curva depende diretamente da taxa de crescimento $ r $. À medida que esse parâmetro aumenta, a inclinação do gráfico torna‑se mais acentuada, indicando um crescimento mais rápido. Isso ocorre porque a taxa de variação da população é proporcional ao número de indivíduos, fazendo com que o crescimento se intensifique ao longo do tempo. Esse padrão evidencia o caráter cumulativo do modelo, no qual o crescimento depende do próprio tamanho da população, de modo que pequenas variações nos parâmetros podem gerar diferenças significativas na dinâmica do sistema.

$ \ \ \ \ \ $ Contudo, apesar de descrever adequadamente situações iniciais ou ambientes com baixa competição, o Modelo de Malthus apresenta limitações importantes. Em sistemas naturais, os recursos são finitos e fatores como competição, predação e disponibilidade de alimento impedem que o crescimento se mantenha indefinidamente.
Dessa forma, o crescimento exponencial é considerado um modelo teórico idealizado, que apesar de sua simplicidade, é importante porque estabelece um limite teórico máximo de crescimento. Ele permite compreender como as populações poderiam crescer na ausência de restrições, servindo como base para modelos mais realistas.

<a id="verhulst"></a>
### 3.2 Modelo de Verhulst

$ \ \ \ \ \ $Na natureza, as populações podem crescer exponencialmente por alguns períodos, mas elas inevitavelmente serão limitadas pela disponibilidade de recursos. O modelo de crescimento logístico, também chamado de Modelo de Verhulst, surge como uma extensão mais realista do crescimento exponencial, incorporando limitações naturais, como alimento, espaço e competição, que impedem que a população cresça indefinidamente.
 
$ \ \ \ \ \ $Nesse modelo, a taxa de crescimento populacional diminui gradualmente à medida que o tamanho da população aumenta. Isso ocorre porque os recursos ambientais são finitos, levando à introdução do conceito de capacidade de suporte $ K $, que representa o tamanho máximo de população que o ambiente pode sustentar ao longo do tempo. [[7]](#1)

O modelo é descrito pela equação diferencial:

 <center> $ \frac{dN}{dt} = rN \left(1 - \frac{N}{K}\right) $  <center> <br>

onde: <br>
$ N(t) $: tamanho da população <br>
$ r $ : taxa de crescimento intrínseca da população <br>
$ K$ : capacidade de suporte do ambiente <br>

O termo:
$$
\left(1 - \frac{N}{K}\right)
$$
representa a limitação ambiental, reduzindo a taxa de crescimento à medida que $ N $ se aproxima de $ K $.

```python
def modelo_logistico(N, t, r, K):
    """
    Modelo de crescimento logístico de uma população (Verhulst).

    Argumentos:
    N : float - Tamanho atual da população no tempo t.
    t : float - Tempo da simulação.
    r : float - Taxa de crescimento intrínseca da população.
    K : float - Capacidade de suporte do ambiente (tamanho máximo da população).
    
    Retorno: float - Taxa de variação da população (dN/dt) no tempo t.
    
    """
    dNdt = r * N * (1 - N / K)
    return dNdt
```

$ \ \ \ \ \ $ Para a obtenção da solução do modelo logístico, utilizou‑se a função odeint, da biblioteca SciPy, que permite resolver numericamente equações diferenciais ordinárias a partir de condições iniciais. A função recebe como entrada o modelo matemático, o valor inicial da população, o vetor de tempo, além dos argumentos opcionais taxa de crescimento $ r $ e capacidade suporte $ K$,  retornando a evolução da população ao longo do tempo.

```python
r = 0.1    
K = 100    
N0 = 10   
     
t = np.linspace(0, 100, 500)

N_logist = odeint(modelo_logistico, N0, t, args=(r, K))
```

$ \ \ \ \ \ $ Essa abordagem possibilita calcular os valores da população em cada instante e, consequentemente, plotar o gráfico da dinâmica populacional, permitindo a visualização do comportamento do sistema.

```python
def plot_cresc_log(r, N0, K):
  
    t = np.linspace(0, 100, 500)

    N_logist = odeint(modelo_logistico, N0, t, args=(r, K))
    
    plt.plot(t, N_logist, color="green", label="Logístico")
    plt.xlabel("Tempo")
    plt.ylabel("População")
    plt.ylim(0, 500)
    plt.title("Modelo de Verhulst")
    plt.axhline(y=K, color="blue", linestyle=":", label="Capacidade de Suporte")
    plt.legend()
    plt.grid()
    plt.show()

interact(
    plot_cresc_log, 
    r = FloatSlider(value=0.1, min=0.001, max=1, step=0.001, description='Taxa de crescimento', style={'description_width':'200px'}, layout=Layout(width='600px')),
    
    N0=FloatSlider(value=10, min=0.5, max=600, step=0.001, description='População Inicial:', style={'description_width':'200px'}, layout=Layout(width='600px')),
    
    K=FloatSlider(value=400, min=0.5, max=500, step=10, description='Capacidade de Suporte:', style={'description_width':'200px'}, layout=Layout(width='600px'))
);
```

$ \ \ \ \ \ $ Assim como o modelo exponencial, o comportamento do modelo logístico depende diretamente dos valores dos parâmetros envolvidos, especialmente da taxa de crescimento $ r $, da população inicial $ N_0 $ e da capacidade de suporte $ K $.

- A taxa de crescimento $ r $ controla a rapidez da dinâmica populacional. À medida que esse parâmetro aumenta, a curva torna-se mais íngreme, indicando que a população cresce mais rapidamente. Isso ocorre porque a taxa de variação da população é proporcional ao número de indivíduos presentes, intensificando o crescimento ao longo do tempo.

- A população inicial $ N_0 $ determina o ponto de partida do sistema. Valores maiores de $ N_0 $ deslocam o início da curva para níveis mais elevados, enquanto valores menores fazem com que a população comece próxima de zero. Embora esse parâmetro altere a posição inicial do gráfico, ele não modifica o valor final de equilíbrio.

-  A capacidade de suporte $ K $ define o limite superior da população no ambiente, atuando como uma assíntota horizontal do sistema. Isso significa que, ao longo do tempo, a população tende a se aproximar de $ K $, sem ultrapassá-lo no modelo ideal.

$ \ \ \ \ \ $ No caso específico em que a população inicial é menor que $ K $, observa-se a forma clássica da curva em S (sigmoide), que pode ser dividida em três fases: uma fase inicial de crescimento aproximadamente exponencial, uma fase de desaceleração devido à limitação de recursos e, por fim, uma fase de estabilização próxima da capacidade de suporte.Já quando a população inicial é maior que a capacidade de suporte $ (N_0 > K) $ , observa-se uma dinâmica oposta ao crescimento inicial: a população diminui ao longo do tempo até se aproximar de $ K $. Esse fenômeno representa a limitação ambiental e a incapacidade do sistema de sustentar populações acima do limite imposto pelos recursos disponíveis.

$ \ \ \ \ \ $ Para compreender melhor a diferença entre os modelos de crescimento exponencial e logístico, foi construído um gráfico que compara os dois comportamentos ao longo do tempo. Essa visualização permite observar como a presença de limitações ambientais — representadas pela resistência do meio (área em verde) — influencia a dinâmica populacional.

$ \ \ \ \ \ $ Para isso, inicialmente foi calculada a solução do modelo exponencial por meio da função `crescimento_exponencial`, que retorna um vetor de tempo $ t $ e o tamanho da população $ N $ ao longo desse intervalo. Em seguida, o modelo logístico foi resolvido numericamente utilizando a função `odeint`, da biblioteca SciPy, que recebe como entrada a equação diferencial, a condição inicial $ N_0 $, o vetor de tempo $ t $ e os parâmetros do modelo, neste caso a taxa de crescimento $ r $ e a capacidade de suporte $ K $.

$ \ \ \ \ \ $ Como o resultado retornado por `odeint` é um array bidimensional, contendo os valores da população para cada instante de tempo, foi necessário selecionar apenas a coluna correspondente à variável populacional, por meio da notação `N_log[:, 0]`, obtendo assim um vetor unidimensional compatível com a plotagem.

$ \ \ \ \ \ $ A partir desses dados, foi possível traçar ambas as curvas no mesmo gráfico, evidenciando que, enquanto o crescimento exponencial não apresenta limitações, o modelo logístico incorpora a resistência do meio, fazendo com que a população se estabilize próxima da capacidade de suporte $ K $, impedindo que a população atinja os níveis previstos pelo crescimento teórico ideal. Contudo, em sistemas reais, a população não necessariamente se mantém exatamente em $ K $, podendo apresentar pequenas oscilações ao redor desse valor devido a variações ambientais, mudanças nos recursos disponíveis e outros fatores ecológicos.

```python
resultado_exp = crescimento_exponencial(r, N0, tf=20)

N_log = odeint(modelo_logistico, N0, t, args=(r, K))
N_log = N_log[:,0]

plt.plot(*resultado_exp, label="Crescimento Exponencial", color="black")
plt.plot(t, N_log, label=" Crescimento Logístico", color="red", linestyle="--")

plt.axhline(y=K, color="blue", linestyle=":", label="Capacidade de Suporte")

plt.fill_between(t, N_log, N_exp, where=(N_exp > N_log),
                 color='green', alpha=0.3, label='Resistência do meio')

plt.xlabel("Tempo")
plt.ylabel("População")
plt.title("Comparação dos modelos")
plt.legend()
plt.grid()
plt.show()
```

### 3.3 Modelo Predador-Presa (Lotka-Volterra)

$ \ \ \ \ \ $ O modelo de Lotka-Volterra descreve matematicamente a interação entre duas espécies: uma população de presas e uma população de predadores. Introduzido independentemente por Alfred Lotka (1925) e Vito Volterra (1926), o modelo surgiu da tentativa de explicar, respectivamente, o observado aumento da população de uma espécie de peixe predador, e a consequente diminuição da população uma espécie de peixe presa, no Mar Adriático durante a Primeira Guerra Mundial, e a variação das concentrações dos elementos químicos que oscilavam em reações químicas. 
Estes modelos, posteriormente denominados modelos de Lotka-Volterra, serviram de base para os modelos matemáticos posteriores utilizados para descrever a dinâmica de sistemas do tipo predador-presa. 

$ \ \ \ \ \ $ O modelo do tipo predador-presa representa um dos primeiros exemplos de sistemas dinâmicos não lineares aplicados à ecologia, sendo amplamente utilizado para compreender a interdependência entre espécies. Atualmente, tais modelos são usados em várias áreas do conhecimento, como em ciências biológicas e agrárias no controle biológico de pragas, em ciências econômicas para descrever as flutuações em bolsas de valores, em ciências ambientais para descrever, por exemplo, a captura e emissão de carbono. A utilização desses modelos permite uma avaliação qualitativa e quantitativa do impacto da competição em diferentes populações sejam elas populações de organismos biológicos, neurônios, bactérias, espécies em ecossistemas e até sistemas econômicos.

$ \ \ \ \ \ $ Para que o modelo funcione, entretanto, ele assume condições idealizadas [[8]](#1):
1. Na ausência de predadores, a presa cresce exponencialmente, isto é, possui recursos infinitos;
2. A morte natural não ocorre;
3. O predador depende exclusivamente daquela espécie de presa para sobreviver, sem ela, o predador morre de fome;
4. A taxa de predação é proporcional ao produto das duas populações $ (xy) $, representando a frequência de encontros aleatórios entre elas;
5. Não há variações sazonais ou genéticas que influenciam as taxas de nascimento e morte;
6. Os predadores se reproduzem em uma taxa proporcional à densidade populacional;
7. Predadores têm apetite ilimitado.

$ \ \ \ \ \ $ Considerando essas hipóteses, seja $x(t)$ a população de presas e $y(t)$ a população de predadores ao longo do tempo. O modelo é descrito pelo seguinte sistema de equações diferenciais:
$$
\frac{dx}{dt} = ax - bxy
$$

$$
\frac{dy}{dt} = -cy + dxy
$$
onde: <br>
$ x(t) $: população de presas <br>
$ y(t) $: população de predadores <br>
$ a $: taxa de crescimento natural das presas <br>
$ b $: taxa de predação <br>
$ c $: taxa de mortalidade dos predadores <br>
$ d $: eficiência da conversão de presas em novos predadores


$ \ \ \ \ \ $ Essas equações evidenciam a interdependência entre as espécies. A população de presas cresce naturalmente, mas é reduzida pela presença de predadores. Por outro lado, a população de predadores diminui na ausência de alimento, mas aumenta quando há disponibilidade de presas. Dessa forma, a partir dessas equações, há três conclusões a serem feitas:
1. Se os predadores se extinguem, as presas crescem exponencialmente, seguindo o Modelo de Malthus:
 $$
   \frac{dx}{dt} = a x \quad \Rightarrow \quad x(t) = x_0 e^{a t}
   $$

2. Se a presa se extinguir, os predadores também desaparecerão logo depois:
$$
   \frac{dy}{dt} = -c y \quad \Rightarrow \quad y(t) = y_0 e^{-c t}
   $$

3.  Se ambas as espécies desaparecerem, o sistema permanece no estado de extinção, não ocorrendo variação ao longo do tempo.

$ \ \ \ \ \ $ A partir de tais noções, é possível criar uma função que simule o Modelo de Lotka-Volterra:

```python
def lotka_volterra(pop, t, a, b, c, d):
    """
    Modelo de Lotka-Volterra para interações predador-presa.

    Argumentos:
    pop: array - Vetor contendo as populações atuais [x, y], onde:
        x = população de presas
        y = população de predadores

    t: float - tempo da simulação.

    a: float - taxa de crescimento natural das presas.
    
    b: float - taxa de predação.
    
    c: float - taxa de mortalidade dos predadores.
    
    d: float - eficiência da conversão de presas em novos predadores
    

    Retorno: list - Lista contendo as derivadas [dx/dt, dy/dt], que representam
        as taxas de variação das populações de presas e predadores.
    """    
    x, y = pop
    
    dxdt = a*x - b*x*y
    dydt = -c*y + d*x*y
    
    return [dxdt, dydt]
```

$ \ \ \ \ \ $ A função `lotka_volterra` recebe como entrada o vetor de populações $pop$, o tempo $t$ e os parâmetros do modelo ($a$, $b$, $c$, $d$). O vetor $pop$ contém dois valores, correspondentes às populações de presas $x(t)$ e predadores $y(t)$, que são atribuídos às variáveis $x$ e $y$ por meio da instrução: `x, y = pop`.

$ \ \ \ \ \ $ Por fim, a função retorna uma lista contendo as duas derivadas:$[dxdt, dydt]$. Esses valores representam a variação instantânea das populações de presas e predadores, sendo utilizados posteriormente pela função odeint (que recebe como entrada a função lotka_volterra, o vetor de condições iniciais, o vetor de tempo $t$ e os parâmetros do modelo $(a, b, c, d)$) para calcular a evolução completa do sistema ao longo do tempo. 

$ \ \ \ \ \ $ Analogamente ao processo realizado para o modelo logístico, a transformação de dados bidimensionais em vetores unidimensionais foi necessária para possibilitar a plotagem das populações de presas e predadores ao longo do tempo. Isso se dá, pois o resultado retornado por odeint é uma matriz bidimensional, na qual cada linha corresponde a um instante de tempo e cada coluna representa uma das variáveis do sistema. Para possibilitar a construção dos gráficos, foi necessário separar esses dados em vetores unidimensionais.

```python
t = np.linspace(0, 50, 1000)

x0 = 10
y0 = 5
iniciais = np.array([x0, y0])

solucao = odeint(lotka_volterra, iniciais, t, args=(a, b, c, d))
```

$ \ \ \ \ \ $ Dessa forma, a população de presas $x(t)$ foi extraída da primeira coluna, enquanto a população de predadores $y(t)$ foi extraída da segunda coluna:

```python
presas = solucao[:, 0]
predadores = solucao[:, 1]
```

$ \ \ \ \ \ $ A partir desses dados, é possível plotar um gráfico que fornece o número de presas em função do tempo:

```python
def plot_presas(x0,y0, a, b, c, d):

    iniciais = np.array([x0, y0])
    solucao = odeint(lotka_volterra, iniciais, t, args=(a, b, c, d))

    presas = solucao[:, 0]
    predadores = solucao[:, 1]

    plt.plot(t, presas, color="orange", label="Presas")
    plt.xlabel("Tempo")
    plt.ylabel("População")
    plt.ylim(0,40)
    plt.title("Lotka-Volterra")
    plt.legend()
    plt.grid()
    plt.show()

interact(
    plot_presas, 
    x0 = FloatSlider(value=10, min=0, max=50, step=1, description='Quantidade Presas Iniciais', style={'description_width':'200px'}, layout=Layout(width='600px')),
    
    y0=FloatSlider(value=5, min=0, max=50, step=1, description='Quantidade Predadores iniciais:', style={'description_width':'200px'}, layout=Layout(width='600px')),
    
    a=FloatSlider(value=1, min=0, max=2, step=0.1, description='Taxa de crescimento das presas:', style={'description_width':'200px'}, layout=Layout(width='600px')),
    
    b=FloatSlider(value=0.1, min=0, max=2, step=0.1, description='Taxa de predação:', style={'description_width':'200px'}, layout=Layout(width='600px')),
    
    c=FloatSlider(value=1.5, min=0, max=2, step=0.1, description='Mortalidade dos predadores:', style={'description_width':'200px'}, layout=Layout(width='600px')),
    
    d=FloatSlider(value=0.075, min=0, max=1, step=0.1, description='Eficiência de conversão:', style={'description_width':'200px'}, layout=Layout(width='600px')),
);
```

$ \ \ \ \ \ $ Sobrepondo ambos os gráficos, para fins de comparação:

```python
def plot_presas_predadores(x0, y0, a, b, c, d):
    iniciais = np.array([x0, y0])
    solucao = odeint(lotka_volterra, iniciais, t, args=(a, b, c, d))

    presas = solucao[:, 0]
    predadores = solucao[:, 1]
    
    plt.plot(t, predadores, label="Predadores")
    plt.plot(t, presas, label="Presas")
    plt.xlabel("Tempo")
    plt.ylabel("População")
    plt.ylim(0,40)
    plt.title("Lotka-Volterra")
    plt.legend()
    plt.grid()
    plt.show()

interact(
    plot_presas_predadores, 
    x0 = FloatSlider(value=10, min=0, max=50, step=1, description='Quantidade Presas Iniciais', style={'description_width':'200px'}, layout=Layout(width='600px')),
    
    y0=FloatSlider(value=5, min=0, max=50, step=1, description='Quantidade Predadores iniciais:', style={'description_width':'200px'}, layout=Layout(width='600px')),
    
    a=FloatSlider(value=1, min=0, max=2, step=0.01, description='Taxa de crescimento das presas:', style={'description_width':'200px'}, layout=Layout(width='600px')),
    
    b=FloatSlider(value=0.1, min=0.1, max=2, step=0.1, description='Taxa de predação:', style={'description_width':'200px'}, layout=Layout(width='600px')),
    
    c=FloatSlider(value=1.5, min=0, max=2, step=0.1, description='Mortalidade dos predadores:', style={'description_width':'200px'}, layout=Layout(width='600px')),
    
    d=FloatSlider(value=0.075, min=0, max=1, step=0.1, description='Eficiência de conversão:', style={'description_width':'200px'}, layout=Layout(width='600px')),
);
```

$ \ \ \ \ \ $ A análise conjunta das populações de presas e predadores permite compreender de forma mais completa a influência de cada parâmetro no comportamento do sistema de Lotka-Volterra. As condições iniciais $x_0$ e $y_0$ determinam o estado inicial do sistema. Valores maiores de $x_0$ aumentam a disponibilidade inicial de alimento, favorecendo o crescimento dos predadores após um curto intervalo de tempo. Por outro lado, valores elevados de $y_0$ intensificam a pressão de predação desde o início, reduzindo rapidamente a população de presas.

$ \ \ \ \ \ $ O parâmetro $a$, taxa de crescimento das presas, controla a velocidade com que essa população cresce na ausência de predadores. Valores mais altos de $a$ resultam em aumentos mais rápidos da população de presas, o que, consequentemente, leva a um crescimento posterior dos predadores. Isso intensifica as oscilações do sistema, aumentando a amplitude dos ciclos.

$ \ \ \ \ \ $ A taxa de predação $b$ determina a intensidade da interação entre as espécies. À medida que $b$ aumenta, a população de presas é reduzida mais rapidamente, o que também impacta a população de predadores, podendo levar a oscilações mais abruptas ou até ao colapso do sistema para valores muito elevados.

$ \ \ \ \ \ $ O parâmetro $c$, taxa de mortalidade dos predadores, atua reduzindo sua população ao longo do tempo. Valores maiores de $c$ dificultam a sobrevivência dos predadores, reduzindo sua presença no sistema e, consequentemente, permitindo que a população de presas cresça mais livremente.

$ \ \ \ \ \ $ Já o parâmetro $d$, que representa a eficiência de conversão de presas em novos predadores, influencia diretamente o crescimento da população de predadores. Valores maiores de $d$ ampliam o impacto da predação, pois tornam a reprodução dos predadores mais eficiente, aumentando a pressão sobre as presas.

$ \ \ \ \ \ $ Como resultado dessas interações, observa-se um comportamento oscilatório característico do modelo, no qual o crescimento das presas precede o aumento dos predadores, seguido por uma redução das presas e, posteriormente, dos predadores. Esse padrão evidencia a interdependência entre as espécies, mostrando que alterações em qualquer parâmetro afetam todo o sistema.

$ \ \ \ \ \ $ Ademais, outro aspecto fundamental desse modelo é a existência de pontos de equilíbrio, nos quais as populações permanecem constantes. Esses pontos de equilíbrio são obtidos impondo que as derivadas das populações em relação ao tempo sejam nulas, ou seja:

$$
\frac{dx}{dt} = 0 \quad \text{e} \quad \frac{dy}{dt} = 0
$$

$ \ \ \ \ \ $ A partir dessa condição, obtém-se o equilíbrio não trivial do sistema: 


$$
x_{eq} = \frac{c}{d}, \quad y_{eq} = \frac{a}{b}
$$

onde:
- $x_{eq}$ representa a população de presas no equilíbrio;
- $y_{eq}$ representa a população de predadores no equilíbrio;
- $a$, $b$, $c$ e $d$ são parâmetros do modelo relacionados às taxas de crescimento, predação, mortalidade e conversão.


$ \ \ \ \ \ $ O ponto de equilíbrio representa uma situação em que há uma quantidade exata de presas suficiente para sustentar a população de predadores, enquanto os predadores controlam o crescimento das presas, estabelecendo um balanço dinâmico.

$ \ \ \ \ \ $ Para visualizar o comportamento do sistema, utiliza-se o diagrama de fase, no qual o eixo horizontal representa a população de presas e o eixo vertical representa a população de predadores. Nesse tipo de gráfico, cada ponto corresponde a um estado possível do sistema (ou seja, uma combinação específica de presas e predadores).
A trajetória plotada no diagrama de fase mostra como o sistema evolui ao longo do tempo. No modelo clássico de Lotka-Volterra, essas trajetórias geralmente formam curvas fechadas ao redor do ponto de equilíbrio, indicando um comportamento oscilatório. Isso significa que as populações de presas e predadores não permanecem constantes, mas variam ciclicamente.

$ \ \ \ \ \ $ O ponto de equilíbrio é destacado graficamente como um ponto no plano de fase (marcado em vermelho). Esse ponto funciona como o centro das oscilações, de modo que, embora o sistema não tenda a permanecer exatamente nele, as trajetórias giram ao seu redor, evidenciando um equilíbrio dinâmico entre as espécies.

$ \ \ \ \ \ $ Diferentemente dos gráficos anteriores, nos quais cada população é representada em função do tempo \(t\), o diagrama de fase representa diretamente a população de predadores em função da população de presas. Para isso, foi utilizada a solução numérica obtida por meio da função `odeint`, a partir da mesma função `lotka_volterra` e das condições iniciais do sistema.

$ \ \ \ \ \ $ A partir da matriz retornada por `odeint`, foram extraídos os vetores correspondentes às populações de presas $ x(t)$ e predadores $ y(t)$. Esses vetores foram então utilizados para construir o gráfico, associando os valores de $x(t)$ ao eixo horizontal e os de $y(t)$ ao eixo vertical:

```python
def plot_diagrama(a, b, c, d):
    iniciais = np.array([x0, y0])
    solucao = odeint(lotka_volterra, iniciais, t, args=(a, b, c, d))

    presas = solucao[:, 0]
    predadores = solucao[:, 1]
    
    x_equilibrio = c / d
    y_equilibrio = a / b
    
    plt.plot(presas, predadores, label="Trajetória do sistema")
    plt.scatter(x_equilibrio, y_equilibrio, color="red", label="Equilíbrio")
    
    plt.xlabel("Presas")
    plt.ylabel("Predadores")
    plt.title("Diagrama de Fase")
    plt.grid()
    plt.legend(loc=[0.7,0.9], fontsize="small")
    plt.show()

interact(
    plot_diagrama, 
    a=FloatSlider(value=1, min=0, max=2, step=0.1, description='Taxa de crescimento das presas:', style={'description_width':'200px'}, layout=Layout(width='600px')),
    
    b=FloatSlider(value=0.4, min=0.1, max=1, step=0.01, description='Taxa de predação:', style={'description_width':'200px'}, layout=Layout(width='600px')),
    
    c=FloatSlider(value=1.5, min=0, max=1.8, step=0.1, description='Mortalidade dos predadores:', style={'description_width':'200px'}, layout=Layout(width='600px')),
    
    d=FloatSlider(value=0.075, min=0.01, max=1, step=0.1, description='Eficiência de conversão:', style={'description_width':'200px'}, layout=Layout(width='600px')),
);
```

$ \ \ \ \ \ $ Variando os parâmetros com os Sliders é perceptível que o parâmetro $a$, que consiste na taxa de crescimento das presas, atua diretamente sobre a posição do equilíbrio na direção vertical. Assim, o aumento de $a$ desloca o ponto de equilíbrio para valores mais altos de predadores, ampliando a órbita no eixo vertical,
intensificando as oscilações, pois aumentam a disponibilidade de alimento para os predadores.

$ \ \ \ \ \ $ A taxa de predação $b$ também influencia a posição do equilíbrio, atuando de forma inversa à taxa $a$. Como $y_{eq} = \frac{a}{b}$, o aumento de $b$ reduz o valor de equilíbrio dos predadores. No diagrama de fase, isso faz com que a órbita se torne mais comprimida verticalmente, pois a população de presas é mais rapidamente reduzida pela ação dos predadores.

$ \ \ \ \ \ $ O parâmetro $c$, que representa a mortalidade dos predadores, influencia a posição do equilíbrio no eixo horizontal, já que $x_{eq} = \frac{c}{d}$. Quanto maior o valor de $c$, maior será o número de presas necessário para sustentar a população de predadores, deslocando o ponto de equilíbrio para a direita. Isso também altera a forma da órbita, aumentando sua extensão no eixo das presas.

$ \ \ \ \ \ $ Já o parâmetro $d$, associado à eficiência de conversão de presas em novos predadores, apresenta efeito oposto ao de $c$. Como $x_{eq} = \frac{c}{d}$, o aumento de $d$ reduz o valor de equilíbrio das presas, deslocando o ponto de equilíbrio para a esquerda. Esse comportamento indica que, quanto mais eficiente for a conversão, menor será a quantidade de presas necessária para sustentar a população de predadores.

$ \ \ \ \ \ $ Dessa forma, o diagrama de fase fornece uma visão global da dinâmica do sistema, evidenciando como a relação entre os parâmetros determina simultaneamente o comportamento das populações de presas e predadores.
