## 1. Introdução

 O estudo das populações constitui uma das bases fundamentais da Ecologia, sendo essencial para compreender como os organismos variam ao longo do tempo e interagem entre si e com o ambiente. A análise da dinâmica populacional permite investigar fenômenos que vão desde o crescimento de uma única espécie até interações complexas em comunidades e ecossistemas, como competição, predação e propagação de doenças. 
 
Dentro desse contexto, modelos matemáticos desempenham um papel central, pois possibilitam descrever, prever e interpretar o comportamento das populações por meio de equações que representam processos ecológicos reais. Modelos clássicos, como o crescimento exponencial e logístico, permitem compreender os efeitos da disponibilidade de recursos e da capacidade de suporte do ambiente, enquanto modelos de interação, como o sistema de Lotka-Volterra, descrevem a relação entre predadores e presas.
    
Além da modelagem, a análise de comunidades ecológicas envolve métricas quantitativas que caracterizam sua estrutura, como riqueza, diversidade e equidade. Índices como o de Shannon-Wiener e o de Pielou são amplamente utilizados para avaliar a distribuição de indivíduos entre espécies e a organização das populações dentro de uma comunidade.

## 2. Fundamentação Teórica

O estudo das populações foca em grupos de indivíduos de uma mesma espécie que coexistem em uma determinada área ou habitat. A variação no tamanho dessas populações ao longo do tempo é regida por quatro processos demográficos fundamentais: nascimentos (B), mortes (D), imigração (I) e emigração (E). Em modelos de populações fechadas, desconsidera-se o fluxo migratório, focando-se na taxa de crescimento intrínseca (r), que resulta da diferença entre as taxas de natalidade e mortalidade.
A modelagem matemática permite descrever como as populações ocupam o ambiente sob diferentes condições de recursos:
Crescimento Exponencial: Proposto inicialmente por Malthus, assume recursos ilimitados onde a taxa de variação da população é proporcional ao seu tamanho atual ( (dN/dt) = rN). Este modelo prevê um crescimento infinito, o que raramente é sustentável na natureza a longo prazo.

Crescimento Logístico: Introduzido por Pierre Verhulst, este modelo considera que o ambiente possui uma capacidade de suporte (K), que representa o limite máximo de indivíduos que os recursos disponíveis podem sustentar. A equação ( (dN/dt) = rN(1-N/K)) descreve um crescimento que desacelera conforme a população se aproxima de K, resultando em uma curva sigmoide (em formato de S).

## 4. Modelagem e Análise Computacional

### 4.1 Crescimento Exponencial

O crescimento populacional exponencial descreve uma situação ideal em que uma população aumenta continuamente a uma taxa proporcional ao seu tamanho. Nesse modelo, a taxa de crescimento por indivíduo (taxa per capita) permanece constante ao longo do tempo, independentemente do tamanho populacional. [2] <br>
Isso implica que, quanto maior a população, maior será o número absoluto de novos indivíduos gerados, levando a um crescimento cada vez mais acelerado. Matematicamente, esse comportamento é representado pela equação diferencial:

$$
\frac{dN}{dt} = rN
$$
onde:<br>
$ N $: representa o tamanho da população <br>
$ r $: é a taxa de crescimento intrínseca <br>

A solução dessa equação é dada por:

$$
N(t) = N_0 e^{rt}
$$

onde $ (N_0) $ é a população inicial. <br>

Vale ressaltar que esse modelo pressupõe condições ideais, como disponibilidade ilimitada de recursos e ausência de competição, sendo considerado uma aproximação válida apenas em estágios iniciais de crescimento populacional, , como na colonização de um novo ambiente ou em culturas de microrganismos em laboratório.

```python
import numpy as np
import matplotlib.pyplot as plt

# Parâmetros
r = 0.1        # taxa de crescimento
N0 = 10        # população inicial

# Tempo
t = np.linspace(0, 100, 500)

# Solução do modelo exponencial
N = N0 * np.exp(r * t)
```

A partir da simulação, foi gerado o gráfico da população ao longo do tempo:

```python
plt.figure()
plt.plot(t, N, color='blue')
plt.xlabel("Tempo")
plt.ylabel("População")
plt.title("Crescimento Exponencial")
plt.grid()
plt.show()
```

Observa‑se que o crescimento populacional aumenta de forma cada vez mais rápida ao longo do tempo, evidenciando o caráter multiplicativo do modelo. Esse comportamento ocorre porque a taxa de crescimento é diretamente proporcional ao número de indivíduos, gerando um efeito cumulativo.

Apesar de descrever adequadamente situações iniciais ou ambientes com baixa competição, o modelo exponencial apresenta limitações importantes. Em sistemas naturais, os recursos são finitos e fatores como competição, predação e disponibilidade de alimento impedem que o crescimento se mantenha indefinidamente.
Dessa forma, o crescimento exponencial é considerado um modelo teórico idealizado, sendo posteriormente ajustado por modelos mais realistas, como o crescimento logístico, que incorpora a limitação ambiental.

No entanto, apesar de sua simplicidade, o modelo exponencial é importante porque estabelece um limite teórico máximo de crescimento. Ele permite compreender como as populações poderiam crescer na ausência de restrições, servindo como base para modelos mais realistas.

### 4.2 Crescimento Logístico

 Na natureza, as populações podem crescer exponencialmente por alguns períodos, mas elas inevitavelmente serão limitadas pela disponibilidade de recursos. O modelo de crescimento logístico surge como uma extensão mais realista do crescimento exponencial, incorporando limitações naturais, como alimento, espaço e competição, que impedem que a população cresça indefinidamente.
 
Nesse modelo, a taxa de crescimento populacional diminui gradualmente à medida que o tamanho da população aumenta. Isso ocorre porque os recursos ambientais são finitos, levando à introdução do conceito de capacidade de suporte $ K $, que representa o tamanho máximo de população que o ambiente pode sustentar ao longo do tempo. [2]

O modelo é descrito pela equação diferencial:

 <center> $ \frac{dN}{dt} = rN \left(1 - \frac{N}{K}\right) $  <center> <br>

onde: <br>
$ N(t) $: tamanho da população <br>
$ r $ : taxa de crescimento intrínseca <br>
$ K$ : capacidade de suporte do ambiente <br>

O termo:
$$
\left(1 - \frac{N}{K}\right)
$$
representa a limitação ambiental, reduzindo a taxa de crescimento à medida que $ N $ se aproxima de $ K $.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint

# Parâmetros
r = 0.1    # taxa de crescimento
K = 100    # capacidade de suporte
N0 = 10    # população inicial

# Tempo
t = np.linspace(0, 100, 500)

# Equação logística
def modelo_logistico(N, t):
    dNdt = r * N * (1 - N / K)
    return dNdt

# Resolvendo a EDO
N_log = odeint(modelo_logistico, N0, t)
```

A partir da simulação, foi gerado o gráfico da população ao longo do tempo:

```python
plt.figure()
plt.plot(t, N_log, color='green', label="Logístico")
plt.xlabel("Tempo")
plt.ylabel("População")
plt.title("Crescimento Logístico")
plt.legend()
plt.grid()
plt.show()
```

A solução dessa equação apresenta uma curva característica em forma de S (sigmoide), na qual o crescimento inicial é aproximadamente exponencial, mas desacelera à medida que a população se aproxima de $ K $. 

O gráfico é composto por três etapas principais:
1. Fase Inicial - Exponencial: Inicialmente, o gráfico se comporta de forma que há um crescimento rápido, baixa competição e recursos abundantes.
2. Fase de Desaceleração: Com o aumento da competição, há uma redução da taxa de crescimento.
3. Fase de Equilíbrio: A população se estabiliza próxima de $ K $, de modo que o crescimento líquido tende a zero.

Para compreender melhor a diferença entre os modelos de crescimento exponencial e logístico, será construído um gráfico que compara os dois comportamentos ao longo do tempo. Essa visualização permite observar como a presença de limitações ambientais influencia a dinâmica populacional.

```python
t = np.linspace(0, 20, 500)
r = 0.2
K = 100
N0 = 10

N_exp = N0 * np.exp(r * t)
N_log = K / (1 + ((K - N0)/N0) * np.exp(-r * t))

plt.figure()

plt.plot(t, N_exp, color='black', label='Crescimento Exponencial')
plt.plot(t, N_log, color='red', label='Crescimento Logístico')

plt.axhline(y=K, color='blue', linestyle='--', label='Capacidade de Suporte (K)')

plt.fill_between(t, N_log, N_exp, where=(N_exp > N_log),
                 color='green', alpha=0.3, label='Resistência do meio')

plt.xlabel("Tempo")
plt.ylabel("Número de indivíduos")
plt.title("Crescimento populacional")

plt.legend()
plt.grid()

plt.show()
```

O gráfico evidencia que o crescimento logístico é uma extensão mais realista do crescimento exponencial. Inicialmente, ambos apresentam comportamento semelhante, pois a população ainda é pequena e os recursos são abundantes. Contudo, à medida que a população cresce, fatores limitantes — como a escassez de alimento, espaço e aumento da competição — reduzem progressivamente a taxa de crescimento.

Diferente do modelo exponencial, o crescimento logístico considera explicitamente a interação com o ambiente, fazendo com que a população se estabilize próxima à capacidade de suporte $ K $, evitando assim o crescimento ilimitado. A diferença entre as curvas exponencial e logística representa a resistência do meio, que impede que a população atinja os níveis previstos pelo crescimento teórico ideal.

Além disso, em sistemas reais, a população não necessariamente se mantém exatamente em $ K $, podendo apresentar pequenas oscilações ao redor desse valor devido a variações ambientais, mudanças nos recursos disponíveis e outros fatores ecológicos.

### 4.3 Modelo Predador-Presa (Lotka-Volterra)

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint

# Parâmetros
a = 1.0    # crescimento das presas
b = 0.1    # predação
c = 1.5    # mortalidade dos predadores
d = 0.075  # reprodução dos predadores

# Função do modelo (EDO)
def modelo(pop, t):
    x, y = pop
    
    dxdt = a*x - b*x*y
    dydt = -c*y + d*x*y
    
    return [dxdt, dydt]

# Condições iniciais
x0 = 10
y0 = 5
iniciais = [x0, y0]

# Tempo
t = np.linspace(0, 50, 1000)

# Resolver sistema
solucao = odeint(modelo, iniciais, t)

# Separar resultados
presas = solucao[:, 0]
predadores = solucao[:, 1]
```

```python
plt.plot(t, presas, label="Presas")
plt.plot(t, predadores, label="Predadores")
plt.xlabel("Tempo")
plt.ylabel("População")
plt.title("Lotka-Volterra")
plt.legend()
plt.grid()
plt.show()
```

```python
plt.plot(presas, predadores)
plt.xlabel("Presas")
plt.ylabel("Predadores")
plt.title("Diagrama de Fase")
plt.grid()
plt.show()
```

```python
x_eq = c / d
y_eq = a / b

plt.plot(presas, predadores)
plt.scatter(x_eq, y_eq, color='red', label="Equilíbrio")
plt.legend()
plt.show()
```

### 4.4 Análise Numérica do Sistema

```python
from scipy.integrate import odeint
import numpy as np
import matplotlib.pyplot as plt

def modelo(pop, t, a, b, c, d):
    x, y = pop
    dxdt = a*x - b*x*y
    dydt = -c*y + d*x*y
    return [dxdt, dydt]

t = np.linspace(0, 50, 1000)
iniciais = [10, 5]

# diferentes valores de b
b_vals = [0.05, 0.1, 0.2]

for b in b_vals:
    sol = odeint(modelo, iniciais, t, args=(a, b, c, d))
    plt.plot(t, sol[:,0], label=f"b={b}")

plt.xlabel("Tempo")
plt.ylabel("Presas")
plt.title("Sensibilidade à taxa de predação")
plt.legend()
plt.show()
```

### 4.5 Índices Ecológicos

Além de analisar a variação do tamanho das populações ao longo do tempo, é importante avaliar a estrutura da comunidade ecológica, considerando como os indivíduos estão distribuídos entre diferentes espécies.
Para isso, são utilizados índices ecológicos que quantificam aspectos como diversidade e equidade. Entre os mais utilizados destacam‑se o Índice de Shannon-Wiener e o Índice de Pielou.

```python
import numpy as np

H_tempo = []
J_tempo = []

for i in range(len(presas)):
    abundancias = np.array([presas[i], predadores[i]])
    
    total = np.sum(abundancias)
    p = abundancias / total
    
    # Remove zeros para evitar log
    p = p[p > 0]
    
    H = -np.sum(p * np.log(p))
    
    S = len(p)
    J = H / np.log(S) if S > 1 else 0
    
    H_tempo.append(H)
    J_tempo.append(J)
```

```python
plt.plot(t, H_tempo)
plt.xlabel("Tempo")
plt.ylabel("Diversidade (Shannon)")
plt.title("Índice de Shannon")
plt.grid()
plt.show()
```

```python
plt.plot(t, J_tempo)
plt.xlabel("Tempo")
plt.ylabel("Equidade (Pielou)")
plt.title("Índice de Pielou")
plt.grid()
plt.show()
```
